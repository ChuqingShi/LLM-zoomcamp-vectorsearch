```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
from sqlitesearch import VectorSearchIndex

vs_index = VectorSearchIndex(
    keyword_fields=["course"],
    mode="ivf",
    db_path="faq_vectors2.db"
)
```

```python
vs_index.clear()
```

```python
from ingest import iter_faq_batches
from embedding import preprocess_documents, get_embedding

batch_size = 32

for batch in iter_faq_batches(batch_size): # generators don't know their length ahead of time
    _, batch_texts = preprocess_documents(batch)
    X = get_embedding(batch_texts, model, batch_size)
    for i in range(len(X)):
        vs_index.add(X[i], batch[i])
```

```python
query_vector = model.encode("How do I run Kafka?")
results = vs_index.search(query_vector, num_results=5)
```

```python
results
```

```python
query = "I missed my homework deadline. What is the consequence?"
query_vector = model.encode(query)

results = vs_index.search(
    query_vector,
    filter_dict={"course": "llm-zoomcamp"},
    num_results=5
)
```

```python
results
```

```python
vs_index.close()
```
