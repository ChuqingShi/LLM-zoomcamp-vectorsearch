```python
from ingest import load_faq_data
documents = load_faq_data()
```

```python
texts = []

for doc in documents:
    text = doc["question"] + " " + doc["answer"]
    texts.append(text)
```

```python
len(texts)
```

```python
from tqdm.auto import tqdm
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
model
```

```python
model.save("./models/all-MiniLM-L6-v2")  # cache the model locally (on Codespace) so we don't have to download it every time
```

```python
batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i:i + batch_size]
    batch_vectors = model.encode(batch)
    vectors.extend(batch_vectors)

len(vectors)
```

```python
import json
with open("documents.json", "w", encoding="utf-8") as f: # utf-8 makes the file portable across OSes
    json.dump(documents, f, ensure_ascii=False, indent=2) # represents Unicode characters readable
```

```python
from ingest import load_faq_data
from embedding import preprocess_documents, get_embedding, vector_search
from sentence_transformers import SentenceTransformer
```

```python
documents = load_faq_data()
documents, texts = preprocess_documents(documents, "llm-zoomcamp")

model = SentenceTransformer('all-MiniLM-L6-v2')
X = get_embedding(texts, model, batch_size=32)

query = "Can I still join the course after the start date?"
results = vector_search(X, documents, query, model, top_k=5)
```
